The cell joins text and a number with `+`. Python does not turn a number into
text on its own, so `"..." + 1841` has no defined meaning and Python stops with a
`TypeError`. Fix it by making the year text. Either wrap the year with `str(...)`,
or build the whole sentence with an f-string, which inserts each value for you,
as in `f"{president} ... {year}"`.

Fix the cell so it prints one sentence and `message` reads
`Harrison gave the longest inaugural address, in 1841`.

````python
year = 1841
president = "Harrison"

message = president + " gave the longest inaugural address, in " + str(year)
print(message)
````

## Task 4. Draw one figure

This checks that plotting works and that figures survive the export to PDF. 

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 3.2))
ax.scatter(speeches["year"], speeches["tokens"], s=20, color="#00693e")
ax.set(xlabel="Year", ylabel="Tokens in the address")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
````

One address towers over every other. That is William Henry Harrison in 1841, who
spoke for close to two hours in the cold without a coat and died thirty-one days
later. The shortest is Washington's second, at 135 tokens.
